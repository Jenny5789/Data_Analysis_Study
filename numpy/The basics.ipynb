# The basics

 **ndim**: 배열의 차원 수 (축의 개수)
 - 예시: 3x4 행렬의 경우 -> 2

**shape**: 각 차원의 크기 (배열의 모양)
 -  예시: 3x4 행렬의 경우 -> (3, 4)

**size**: 배열 내 전체 요소의 총 개수
 - 예시: 3x4 행렬의 경우 -> 12

**dtype**: 배열에 저장된 요소의 자료형
 - 예시: 정수형일 경우 -> int64
 
**itemsize**: 요소 하나가 차지하는 바이트(byte) 크기
 - 예시: int64 타입일 경우 -> 8
 
**data**: 실제 데이터가 저장된 메모리 버퍼
 - 예시: 직접 접근하기보다는 인덱싱을 통해 주로 사용

## Basic operations

```python
a = np.array([20, 30, 40, 50])
b = np.arange(4)
b
```

```python
c = a - b
c   # 같은 위치(index)에 있는 요소끼리 연산 수행됨  -> 요소별 연산
```

```python
b ** 2   # b_squared = b^2(제곱) 의미
# b ^ 2: 비트 단위 XOR 연산(두 비트가 다를 때만 1 반환)(수학적 제곱 아님)
```

```python
10 * np.sin(a)  # 배열 a의 각 요소에 sin 값을 구하고 10을 곱함 (요소별 연산)
```

```python
a < 35  # a의 각 요소가 35보다 작은지 판별하여 불리언(True/False) 배열 반환
```

```python
# [브로드캐스팅]
# 크기가 다른 배열끼리 연산 시, 작은 배열을 자동으로 확장하여 크기를 맞춤
# 예: [1, 2, 3] + 10  -> [1+10, 2+10, 3+10]
```

```python
A = np.array([[1, 1],
              [0, 1]])
B = np.array([[2, 0],
              [3, 4]])
A * B   # 같은 위치에 있는 요소끼리 단순히 곱. (행렬의 수학적 곱셈이 아님, 1:1 대응 곱셈)
```

```python
A @ B  # 선형대수학에서 정의하는 표준 행렬 곱셈, 데이터 간의 관계를 변환하거나 조합할 떄 사용하는 행렬곱셈
```

```python
A.dot(B)   #A @ B와 동일한 결과(행렬의 곱셈)
```

## Indexing, Slicing and iterating

```python
a = np.arange(10) ** 3   # 0 ~ 9까지 1차원 배열 생성, 각 요소를 3제곱(cybe) 처리
a
```

```python
a[2] # 인덱스 2번 '값' 하나를 추츨
```

```python
a[2:5] # 인덱스 2번 부터 4번까지 '범위' 출력
```

```python
a[:6:2] = 1000    # 인덱스 0부터 6이전까지 2씩 건너뛰며 접근하여 '1000'으로 일괄 변경
a
```

```python
a[::-1]  # 시작부터 끝까지 뒤에서 -1칸씩 이동
```

```python
for i in a:
    print(i ** (1/3.))

# [역정규화(Denormalization) 및 복원]  'i ** (1/3.)'
# - 목적: 세제곱(x**3) 연산의 역연산으로, 원래의 데이터 단위로 '역정규화(복원)' 수행
# - 수학적 의미: i^(1/3) = 세제곱근(Cube Root)
# - 참고: 부동소수점 연산 특성상 완벽한 정수가 아닌 미세한 오차가 발생할 수 있음
#        (9.999...와 같이 출력될 경우 np.round()로 보정 가능)
```

```python
def f(x, y):              # x: 행 인덱스 값, y: 열 인덱스 값 (좌표)
    return 10 * x + y
```

```python
b = np.fromfunction(f, (5, 4), dtype=np.int_)
b

# [np.fromfunction]
# - 함수(x, y)의 연산 결과값들을 받아 5행 4열 2차원 배열 생성
# - int_(현재 시스템의 기본 정수 타입)의 type 형태로 지정
```

```python
b[2, 3]  # 스칼라 인덱싱, 배열의 type 정보, 배열의 특정 위치(2행 3열)의 값 하나를 직접 추출

# 스칼라(Scalar, 단일 값, 차원이 없는 값) 인덱싱: '값' 하나를 가져오며, 그 값의 타입 정체성을 명확히 표기
```

```python
b[0:5, 1]  # 슬라이싱, 값 하나가 아니라 '0행부터 4행까지, 1열의 데이터 범위'을 추출(부분 집합 추출)

# 슬라이싱(Slicing, 부분 집합): 값 하나가 아닌 배열의 일부분을 구조적으로 추출하며, 결과물은 여전히 '배열(Array)' 형태
```

```python
b[:,1]    # 모든 행(:)에 대하여 1열의 데이터를 가져옴
```

```python
b[1:3, :] # 1행부터 2행까지 전체 열(:)을 가져옴
```

```python
# [인덱싱 및 슬라이싱 구분자 : 쉼표(,)]
# - 쉼표 기준 앞: 행(Row) 범위 지정///쉼표 기준 뒤: 열(Column) 범위 지정

# b[:, :]: 모든 행(:)과 모든 열(:)을 선택 -> 결과: 배열 전체
# b[0, :]: 0행의 모든 열 (1차원 배열 반환)
# b[:, 0]: 모든 행의 0열 (1차원 배열 반환)
# b[0, 0]: 0행 0열의 단일 값 (스칼라 반환)
```

```python
b[-1]  # 뒤에서 첫 번째 행을 가져옴
```

```python
c = np.array([[[0, 1, 2],[10,12,13]],
              [[100,101,102],[110,112,113]]])
c.shape
```

```python
# [Ellipsis(...) 활용]
# - 원칙: 나머지 모든 차원을 '전부' 선택하며, 괄호 내 단 1회만 사용, ... 은 '생략된 차원의 뭉치'입니다.
# - 구조적 의미:c[층, 행, 열] 기준
#    c[1, ...]  → c[1, :, :]  (1번 층 고정, 나머지 전부)
#    c[..., 2]  → c[:, :, 2]  (2번 열 고정, 나머지 전부)
#    c[..., 1, ...] 과 같이 작성하면, 어느 쪽 ...이 어디까지 담당할지 모호해집니다.
```

```python
c[1,...]   # 1번 층의 모든 행/열 슬라이싱 
# 구조: c[1, :, :]  -> 문법: c[1, ...]
```

```python
c[...,2]  # 모든 층/행의 2열 인덱싱  c[..., ..., 2]
# 구조: c[:, :, 2]  -> 문법: c[..., 2]
```

```python
for row in b:
    print(row)

# [for row in b]
# 행 단위로 순회, (구조 유지)
```

```python
for element in b.flat:
    print(element)

# [for element in b.flat] 
# 개별 요소 단위로 순회, 개별 데이터(스칼라)단위로 출력 (평탄화, 1차원화)
```
