# Advanced indexing and index tricks

## Indexing with arrays of indices

```python
import numpy as np
a = np.arange(12)**2             # 0 ~ 11까지 각 원소를 제곱한 배열 생성
i = np.array([1, 1, 3, 8, 5])    # a에서 꺼낼 인덱스 번호 배열
a[i]                             # i의 값을 인덱스로 사용해 a에서 추출 
```

```python
j = np.array([[3, 4], [9, 7]])  # 2행 2열 배열 생성 (a의 인덱스 번호)
a[j]                             # j의 값을 인덱스로 사용해 a에서 꺼냄 
# j가 2차원이므로 결과도 2차원으로 출력
```

```python
palette = np.array([[0, 0, 0],[255, 0, 0],[0, 255, 0],[0, 0, 255],[255, 255, 255]]) # 5행 3열 배열 생성
 # 0번 : 검정 (RGB), 1번 : 빨강, 2번 : 초록, 3번 : 파랑,  4번 : 흰색
image = np.array([[0, 1, 2, 0],[0, 3, 4, 0]]) # palette 인덱스 번호로 구성된 2행 4열 배열 생성
palette[image]                                #image의 인덱스 번호 이용하여  palette에서 추출함
```

```python
a =np.arange(12).reshape(3, 4)  # 0~11까지 1차원 배열 생성 후 3행 4열로 재배치
a
```

```python
i = np.array([[0, 1],         # 행 인덱스 번호 배열
              [1, 2]])
j = np.array([[2, 1],         # 열 인덱스 번호 배열 
              [3, 3]])
a[i, j]                       # i=행, j=열 인덱스로 a에서 값 추출
```

```python
a[i, 2]       # i행 2열 고정-> i형태 추출(2행2열)
```

```python
a[:, j]      # 모든 행(3개), j열의 인덱스 값 추출
# 3개 묶음 , 2행 2열 형태 반복
```

```python
l = (i, j)       # l은 i와 j로 묶은 튜플,  
# (행,열) 쌍으로 해석, i행 j열 꺼내줘
a[1]             # a 1행 전체
```

```python
s = np.array([i, j])     #i행 j열로 이루어진 배열
# a[s]     s덩어리 배열(인덱스)로 꺼내줘.
```

```python
a[tuple(s)]   # s를 튜플로 변환 -> (행, 열)쌍으로 해석 -> a[i,j]와 동일한 결과
```

```python
time = np.linspace(20, 145, 5)               # 20 ~ 145 사이 5개의 값 생성
data = np.sin(np.arange(20)).reshape(5, 4)   # 0 ~ 19까지의 정수를 sin 함수에 적용 후 5행 4열 형태로 변형
time
# np.linspace(start,stop,num,...) : 시작 값, 끝 값, 생성할 수의 개수
# np.arange(start, stop, step...) : 시작 값, 끝 값(미포함!), 수 사이 간격
```

```python
data
```

```python
ind = data.argmax(axis=0)   # 행 방향으로 최댓값의 인덱스 반환
ind

# 'axis=0'-> 행 방향,  'axis=1'-> 열 방향
```

```python
# times corresponding to the maxima
time_max = time[ind]
data_max = data[ind, range(data.shape[1])]
time_max
```

```python
data_max
```

```python
np.all(data_max == data.max(axis=0))
```

```python
import numpy as np
import matplotlib.pyplot as plt
def mandelbrot(h, w, maxit = 20, r =2):
    "Returns an image of the Mandelbrot fractal of size (h,w)."""
    x = np.linspace(-2.5, 1.5, 4*h+1)
    y = np.linspace(-1.5, 1.5, 3*w+1)
    A, B = np.meshgrid(x, y)
    C = A + B*1j
    z = np.zeros_like(C)
    divtime = maxit + np.zeros(z.shape, dtype=int)

    for i in range(maxit):
        z = z**2 + C
        diverge = abs(z) > r
        div_now = diverge & (divtime == maxit)
        divtime[div_now] = i
        z[diverge] = r

    return divtime
plt.clf()
plt.imshow(mandelbrot(400, 400))
```

## The ix_() function

```python
a = np.array([2, 3, 4, 5])
b = np.array([8, 5, 4])
c = np.array([5, 4, 6, 8, 3])
ax, bx, cx = np.ix_(a, b, c)
ax
```

```python
bx
```

```python
cx
```

```python
ax.shape, bx.shape, cx.shape
```

```python
result = ax + bx * cx
result
```

```python
result[3, 2, 4]
```

```python
a[3] + b[2] * c[4]
```

```python
def ufunc_reduce(ufct, *vectors):
    vs = np.ix_(*vectors)
    r = ufct.identity
    for v in vs:
        r = ufct(r, v)
    return r
```

```python
ufunc_reduce(np.add, a, b, c)
```
