# Shape manipulation

## Changing the shape of an array

```python
import numpy as np
rg = np.random.default_rng(1)     # seed = 1, 난수 생성기 인스턴스 생성
a = np.floor(10 * rg.random((3, 4)))  # 3행 4열 배열로 생성, 10배 후 정수 만듦.(0 ~ 9 정수 배열)
a
```

```python
a.shape   # a의 모양(행, 열) 반환
```

```python
a.ravel()    # a 배열을 1차원으로 펼침 (원본유지)
```

```python
a.reshape(6, 2)  # a를 6행 2열로 재배열 (원본유지)
```

```python
a.T  # a의 전치행렬(행 ↔ 열 교환)
```

```python
a.T.shape # a의 전치행렬의 모양(행, 열) 반환
```

```python
a.shape # a의 모양(행, 열) 반환
```

```python
a
```

```python
a.resize((2, 6))    # 2행 6열로 재배열(원본 직접 변경)
a
```

```python
a.reshape(3, -1)  # a를 3행 x열(자동계산)로 재배열 (원본유지)
```

```python
a = np.floor(10 * rg.random((2, 2)))    # a = 2행 2열 배열로 생성, 10배 후 정수 만듦.(0 ~ 9 정수 배열)
a
```

```python
b = np.floor(10 * rg.random((2, 2)))  # b = 2행 2열 배열로 생성, 10배 후 정수 만듦.(0 ~ 9 정수 배열)
b
```

```python
np.vstack((a, b))  # a, b 행 방향으로 연결하여 작성 (세로 방향)
```

```python
np.hstack((a, b)) # a, b 열방향으로 연결하여 작성 (가로 방향)
```

```python
from numpy import newaxis
np.column_stack((a, b))   # a, b를 열 방향으로 연결
```

```python
a = np.array([4., 2.])
b = np.array([3., 8.])
np.column_stack((a, b))   # a, b를 열벡터로 변환 후 연결 -> shape(2, 2)
```

```python
np.hstack((a, b)) # a, b를 행 방향으로 연결 -> shape(4,)
```

```python
a[:, newaxis]    # a를 열벡터로 변환 (shape (2,) -> (2, 1))
```

```python
np.column_stack((a[:,newaxis], b[:, newaxis]))
# a, b 를 각각 열벡터로 변환
# 열 방향으로 연결-> shape (2, 2)
```

```python
np.hstack((a[:, newaxis], b[:, newaxis]))
# a, b 를 각각 열벡터로 변환
# 행 방향으로 연결-> shape (2, 2)
```

```python
np.r_[1:4, 0, 4]  # '1:4' = 'range(1,4)', row 방향으로 이어 적기
```

## Splitting one array into several smaller ones

```python
a = np.floor(10 * rg.random((2, 12)))   # a = 2행 12열 배열로 생성, 10배 후 정수 만듦.(0 ~ 9 정수 배열)
a 
```

```python
np.hsplit(a, (3, 4))  # a를 열 방향(세로)으로 분할---> 0:3, 3:4, 4:x(분할 위치 지정, ':' 이후 인덱스 미포함)
```
