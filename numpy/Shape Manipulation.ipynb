# Shape manipulation

## Changing the shape of an array

```python
import numpy as np
rg = np.random.default_rng(1)
a = np.floor(10 * rg.random((3, 4)))
a
```

```python
a.shape
```

```python
a.ravel()    # returns the array, flattened
```

```python
a.reshape(6, 2)
```

```python
a.T  #returns the array, transposed
```

```python
a.T.shape
```

```python
a.shape
```

```python
a
```

```python
a.resize((2, 6))
a
```

```python
a.reshape(3, -1)
```

```python
a = np.floor(10 * rg.random((2, 2)))
a
```

```python
b = np.floor(10 * rg.random((2, 2)))
b
```

```python
np.vstack((a, b))
```

```python
np.hstack((a, b))
```

```python
from numpy import newaxis
np.column_stack((a, b))
```

```python
a = np.array([4., 2.])
b = np.array([3., 8.])
np.column_stack((a, b))
```

```python
np.hstack((a, b))
```

```python
a[:, newaxis]
```

```python
np.column_stack((a[:,newaxis], b[:, newaxis]))
```

```python
np.hstack((a[:, newaxis], b[:, newaxis]))
```

```python
np.r_[1:4, 0,4]
```
