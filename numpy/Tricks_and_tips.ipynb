## "Automatic" reshaping

```python
import numpy as np
a = np.arange(30)             # 0 ~ 29(30개)의 원소
b = a.reshape((2, -1, 3))     # 축0(2개) * 축1(미정) * 축2(3개)
# -1은 자동 계산 → 축1 = 5    -->  결과: (2, 5, 3)    ( 2개의 5행 3열 배열)
b.shape
```

## Vector stacking

```python
x = np.arange(0, 10, 2)     # 0 ~ 9까지의 수를 2간격으로 배열
y = np.arange(5)            # 0 ~ 4까지의 수를 배열
m = np.vstack([x, y])       # x, y의 수를 수직으로 쌓기( 두 배열을 행으로 아래 추가)
m
```

```python
xy = np.hstack([x, y])      # x, y의 수를 수평으로 쌓기( 두 배열을 열로 옆에 추가)
xy
```

## Histograms

```python
import numpy as np
rg = np.random.default_rng(1)               # 난수 생성기 초기화 (seed=1로 고정, 매번 같은 난수가 생성)
import matplotlib.pyplot as plt
mu, sigma = 2, 0.5                          # 평균(mean)=2, 표준편차(std)=0.5
v = rg.normal(mu, sigma, 10000)             # 정규분포를 따르는 난수 10000개 생성
                                            # 평균 2 주변에 대부분 모여있는 데이터 생성
plt.hist(v, bins = 50, density = True)      # 히스토그램 그리기(각 구간이 명확)
                                            # 50개 구간으로 범위 나눔, 정규화 면적(=히스토그램 너비 1)
```

```python
# Compute the histogram with numpy and then plot it
(n, bins) = np.histogram(v, bins = 50, density = True)
# np.histogram: 히스토그램 데이터 계산 (그래프 표시 X)
# n: 50개 구간의 높이,   bins : 51개의 구간 경계값

plt.plot(.5 * (bins[1:] + bins[:-1]), n)
# .5  중점 구하기 위함
# bins[1:] + bins[:-1]: 이웃한 경계값들의 합
# x축: 각 구간의 중점 (50개)
# y축: 각 구간의 높이 n (50개)
# 결과: 점들을 선으로 연결한 선 그래프 표시
```
