## An example

```python
import numpy as np
a = np.arange(15).reshape(3, 5)    # 15개의 숫자를 생성하여 3행 5열의 2차원 배열로 구조화
a
```

```python
a.shape   # 배열의 형태, 3행 5열의 구조를 가지고 있음
```

```python
a.ndim       # 배열의 축(axis)의 개수, 행과 열(2개의 축) 2차원
```

```python
a.dtype.name  # 데이터 타입의 이름, 배열 요소들의 자료형(64비트의 정수형)
```

```python
a.itemsize   # 요소당 크기, 배열의 각 요소 하나가 메모리에서 차지하는 크기(8바이트 = 64비트)
```

```python
type(a)  # 배열의 객체 타입, 넘파이의 다차원 배열 객체(ndarray)임
```

## Basic operations

```python
a = np.array([20, 30, 40, 50])
b = np.arange(4)
b
```

```python
c = a - b
c   # 같은 위치(index)에 있는 요소끼리 연산 수행됨  -> 요소별 연산
```

```python
b ** 2   # b_squared = b ** 2
```

```python
10 * np.sin(a)  # 배열 a의 각 요소에 sin 값을 구하고 10을 곱함 (요소별 연산)
```

```python
a < 35  # a의 각 요소가 35보다 작은지 판별하여 불리언(True/False) 배열 반환
```

```python
# [브로드캐스팅]
# 크기가 다른 배열끼리 연산 시, 작은 배열을 자동으로 확장하여 크기를 맞춤
# 예: [1, 2, 3] + 10  -> [1+10, 2+10, 3+10]
```

```python
A = np.array([[1, 1],
              [0, 1]])
B = np.array([[2, 0],
              [3, 4]])
A * B   # 같은 위치에 있는 요소끼리 단순히 곱. (행렬의 수학적 곱셈이 아님)
```

```python
A @ B  # 선형대수학에서 정의하는 표준 행렬 곱셈
```

```python
A.dot(B)   #A @ B와 동일한 결과
```

```python
rg = np.random.default_rng(1)        # 난수 생성기 인스턴스 생성
a = np.ones((2, 3), dtype = int)     # 2행 3열의 모든 요소가 1인 정수형 배열 생성
b = rg.random((2, 3))                # 0 이상 1미만의 난수로 구성된 2행 3열 배열 생성
a *= 3                               # 배열 a의 모든 요소에 3을 곱함
a
```

```python
b += a    # b = float, a = int   -> b = b + a  -> float = float + int 가능!
b
```

```python
a += b    # a = int  b = float   -> a = a + b  -> int = int + float 불가!
```

```python
b = np.arange(12).reshape(3, 4)   # 3행 4열의 2차원 배열 생성
b   
```

```python
b.sum(axis = 0)       # axis = 0, 세로 방향 연산 수행-> 열(col)을 따라 합계 계산
```

```python
b.min(axis = 1)     # axis = 1, 가로 방향 연산 수행-> 행(row)을 따라 최솟값 계산
```

```python
b.cumsum(axis = 1) # axis = 1, 가로 방향 연산 수행-> 행(row)을 따라 누적 합계 계산
```
